# Task 3 — No history, and no single truth

## A. No history — "Where did Ali's sales go?"
**Inmon property broken: Time-variant.** A warehouse must keep data identified with the period it was true for. The live table keeps one `Branch` value and overwrites it, so Ali's March move rewrites every earlier sale as Karachi's; the report can no longer say where he worked at the time of each sale.

**What the warehouse does differently:** it never overwrites the row. When Ali moves, the old row is closed (end date = day before the move) and a **new row** for Ali–Karachi is added with a start date. Each sale is linked to the row that was valid on its sale date, so last year's sales still credit Lahore (a Type 2 slowly changing dimension).

**Another situation where overwriting rewrites history:** a customer's city/tax region or a product's price/category. Overwrite it and old invoices re-total at the new price or land in the new category.

## B. No single truth — Rs 661.65 m vs 283.70 m vs 584.75 m
**One plausible reason each (nobody lying, no arithmetic error):**
- **Operations — Rs 661.65 m:** counts all orders placed, gross, including cancelled and later-returned ones, by order date.
- **Finance — Rs 283.70 m:** counts only recognised revenue — invoiced and paid, net of returns and discounts, by invoice date.
- **Sales — Rs 584.75 m:** counts bookings/deals closed (including orders not yet delivered or invoiced), by deal date.

**What a single staging area changes:** all departments draw from one integrated, cleaned copy where each source is loaded once and the definitions are applied once. The three figures then become **named measures from the same data** (e.g. orders placed, bookings, recognised revenue) and the gaps become explainable adjustments (cancellations, returns, timing) instead of three unreconcilable numbers.

**Who should own the definition:** a business-level data owner/governance group with Finance as the authority on "sales" (revenue), with Operations and Sales as consulted stakeholders — someone accountable, outside any one team's incentives. Not "each keeps its own", because then no two reports can be compared, decisions contradict each other, and the debate is about whose number is right instead of what to do. Departments can still keep their own view, but under a different, explicit name.

## C. "We already have all this data in our systems — why build a warehouse?"
Having the data is not the same as being able to use it. Our own systems showed two failures: the live Employee table **overwrites** values, so Ali's history was moved to Karachi and last year's Lahore report became wrong — the data exists but the *past* does not; and each department's system applies its own definition of "sales", so we got three numbers (Rs 661.65 m, 283.70 m, 584.75 m) for one month and no one could say which to trust. A warehouse keeps history by adding rows instead of overwriting them, and integrates all sources through one staging area with one agreed definition, so the answer is both historically correct and the same for everyone.

## D. Non-volatile
**Definition:** once data is loaded into the warehouse, it is not updated or deleted by day-to-day transactions; changes in the real world are added as new records, so earlier states remain available and users can only read it.

**Example of data a warehouse legitimately changes:** correcting a load error — e.g. a product loaded under the wrong category is fixed by the controlled ETL process (or rows past the retention period are archived/purged) — a deliberate, audited change, not an operational overwrite.
